```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow import keras
```

- Koristeci biblioteku pandas ucitati datotetku fruit.csv
- Ispisati prvih pet instanci skupa podataka

```python
df = pd.read_csv('fruit.csv')
df.head()
```

- Izdvojiti kolonu Class kao ciljnu promenljivu i kodirati je pomocu keras biblioteke

```python
print(df['fruit_label'].unique())
```

```python
classes = df['fruit_label']
classes_cat = keras.utils.to_categorical(classes)
```

- Podeliti podatke na skup za treniranje i skup za testiranje

```python
df = pd.get_dummies(df, prefix='', prefix_sep='')
train_dataset = df.sample(frac=0.8, random_state=42)
test_dataset = df.drop(train_dataset.index)
```

```python
print(train_dataset.shape)
print(test_dataset.shape)
```

```python
X_train = train_dataset.drop('fruit_label', axis=1)
y_train = train_dataset['fruit_label']

X_test = test_dataset.drop('fruit_label', axis=1)
y_test = test_dataset['fruit_label']
```

- Standardizovati podatke

```python
X_train
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
print(len(X_train[0]))
print(len(y_train.unique()))
```

```python
model = keras.models.Sequential()
model.add(keras.layers.Dense(len(X_train[0]), activation='relu'))
model.add(keras.layers.Dense(32, activation='relu'))
model.add(keras.layers.Dense(10, activation='relu'))
model.add(keras.layers.Dense(y_train.unique(), activation='softmax'))
```

```python
print(type(X_train))
print(type(y_train))
```

```python
y_train = keras.utils.to_categorical(y_train.values)
y_test = keras.utils.to_categorical(y_test.values)
print(type(y_train))
```

```python
learning_rates = [0.001, 0.01, 0.1]
best_lr = 0
best_acc = -1.0
best_history = None

for lr in learning_rates:
    model.compile(optimizer=keras.optimizers.Adam(learning_rate=lr), loss=keras.losses.CategoricalCrossentropy(), metrics=['accuracy'])
    history = model.fit(X_train, y_train, epochs = 30)
    _, acc = model.evaluate(X_test, y_test, verbose=0)
    if acc > best_acc:
        best_acc = acc
        best_lr = lr
        best_history = history

print(f'Najbolji learning rate je {best_lr}')
```
