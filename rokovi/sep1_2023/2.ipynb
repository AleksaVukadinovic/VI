```python
from z3 import *
```

```python
# Sorta i promenljive
C = DeclareSort('Cvor')
x, y, z = Consts('x y z', C)

s = Function('s', C, C, BoolSort())   # s(u,v) - u je sused v
p = Function('p', C, C, BoolSort())   # p(u,v) - postoji put od u do v

a, b, c, d, e, f, g, h, m, k = Consts('a b c d e f g h m k', C)

solver = Solver()
axioms = [
    ForAll([x, y], s(x, y) == s(y, x)), # Ako je X sused od Y, onda je i Y sused od X
    ForAll([x, y], p(x, y) == p(y, x)), # Ako postoji put od X do Y, onda postoji i put od Y do X
    ForAll([x, y], p(x, y) == Or(s(x, y), Exists(z, And(s(z, y), p(x, z))))) # p(x,y) <=> s(x,y) OR (exists z. s(z,y) & p(x,z))
]
solver.add(axioms)

# a - b - c
solver.add(s(a, b))
solver.add(s(b, c))
# a - d - e - f
solver.add(s(a, d))
solver.add(s(d, e))
solver.add(s(e, f))
# d - g
solver.add(s(d, g))
# e - h - m - k
solver.add(s(e, h))
solver.add(s(h, m))
solver.add(s(m, k))

solver.push()
solver.add(Not(p(a, f)))
res = solver.check()
if res == unsat:
    print("Dokazano: postoji put od a do f")
else:
    print("Ne može se dokazati da postoji put od a do f")
    if res == sat:
        print("Protivprimer:")
        print(solver.model())
solver.pop()
```
